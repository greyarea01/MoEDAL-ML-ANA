# Efficiency and Purity Calculation for CNN Front Pits

This script calculates the efficiency and purity of predicted etch pits from a CNN model (front exposure) using a true etch pit dataset. It creates two plots: one for efficiency and another for purity, both with respect to varying cut-off distances between the true and predicted pits.

## Key Components

1. **Efficiency and Purity Definitions:**
   - **True Positive (A):** Etch pits that exist both in the clean dataset and the exposed dataset within a certain distance.
   - **False Negative (B):** Etch pits that exist in the clean dataset but not in the exposed dataset within a certain distance.
   - **False Positive (C):** Etch pits that exist in the exposed dataset but not in the clean dataset within a certain distance.

2. **Efficiency and Purity Formulas:**
   - **Efficiency:** \( \text{Efficiency} = \frac{A}{A + B} \)
   - **Purity:** \( \text{Purity} = \frac{A}{A + C} \)

3. **Main Functionality:**
   - The script computes the distances between the predicted and true etch pits.
   - It then calculates the number of true positives, false negatives, and false positives for varying cut-off distances.
   - The efficiency and purity are calculated and plotted as functions of these cut-off distances.

## Functions

### `get_abc(distance_true, distance_pred)`
- **Description:** Calculates the A, B, and C values based on the nearest distances between predicted and true pits.
- **Parameters:**
  - `distance_true`: The nearest distance from each true etch pit to the predicted pits.
  - `distance_pred`: The nearest distance from each predicted etch pit to the true pits.
- **Returns:** Lists `a`, `b`, and `c` representing true positives, false negatives, and false positives respectively.

### `get_eff_pur(a, b, c)`
- **Description:** Calculates efficiency and purity based on the A, B, and C values.
- **Parameters:**
  - `a`: True positives.
  - `b`: False negatives.
  - `c`: False positives.
- **Returns:** Two lists: `efficiency` and `purity` representing the curves with respect to the cut-off distances.

## Output

The script generates a plot with:
- **Efficiency** (orange curve) and **Purity** (blue curve) vs. cut-off distance.
- The plot is saved as `eff_pur_CNN.png` in the `plots/` directory.
